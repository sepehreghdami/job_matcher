```python
from sentence_transformers import SentenceTransformer
import numpy as np

# Load embedding model (runs fine on CPU or your RX 480 via CUDA/ROCm if available)
model = SentenceTransformer("BAAI/bge-base-en-v1.5")

resume_text = """
Backend software engineer, 3 years experience. Kotlin/Ktor microservices,
domain-driven design, clean architecture. PostgreSQL, Exposed ORM, Koin,
Gradle multi-module. AI/ML: multi-agent RAG systems, LangGraph, LangChain,
production LLM integrations.
"""

job_post = """
AI Developer

🏢 unfold ai
⌨️ #Programming
🏷️ #python #react #vue #postgresql #ai #aws #cloud #tensorflow #pytorch #flutter

Fully Remote ✅

📄 Design, implement, and scale modular AI agents to process wearable device data and deliver personalized health insights. Develop backend systems for API integration, data normalization, and machine learning model deployment. Collaborate with cross-functional teams to build scalable architecture and …

Over 700 new job opportunities are posted every day on Jaabz.com.

📝 Apply Now
🌐 View All Jobs
"""

# bge models recommend a query prefix for retrieval tasks
resume_emb = model.encode(f"Represent this resume for matching job posts: {mycvtext}", normalize_embeddings=True)
job_emb = model.encode(job_post_text, normalize_embeddings=True)

similarity = float(np.dot(resume_emb, job_emb))  # cosine similarity, range -1 to 1
print(f"Similarity score: {similarity:.4f}")
```
